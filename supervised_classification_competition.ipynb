```python
def info(df):
    inf=pd.DataFrame({
        'dtype':df.dtypes.astype(str),
        'NaN':df.isna().sum(),
        'missing %':(df.isna().mean()*100).round(2),
        'unique':df.nunique()
    })
    return inf
```
