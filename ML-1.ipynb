```python
import seaborn as sns
df=sns.load_dataset('Titanic')
import warnings
warnings.filterwarnings('ignore')
```

```python
df
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer

def age_(data):
    imp=SimpleImputer(strategy= 'mean')
    data['age']=imp.fit_transform(data[['age']])
    return data

df=age_(df)
```

```python
df
```

```python
cat_imp=SimpleImputer(strategy='most_frequent')

df['embark_town']=cat_imp.fit_transform(df[['embark_town']]).ravel()
```

```python


def preprocess(data):
    data['class']=data['class'].map({
        'Third':1,
        'Second':2,
        'First':3
    })
    data['who']=data['who'].map({
        'child':3,
        'woman':2,
        'man':1
    })
    data['adult_male']=data['adult_male'].map({
        True:1,
        False:0
    })
    return data

df=preprocess(df)
```

```python
df
```

```python
x=df.drop(['survived','deck','embark_town','alive','embarked','sex'],axis=1)
y=df['survived']

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
